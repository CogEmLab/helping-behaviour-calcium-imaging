```python
import pandas as pd
from glob import glob
import os
import numpy as np
```

```python
session_file_path = 'data/updated_session_file_MiniscopeCW_all_Moises.csv'
session_file = pd.read_csv(session_file_path)
# Rename Session outcome to Session Outcome
session_file.rename(columns={'Session outcome': 'Session Outcome'}, inplace=True)
# Convert Cohort to string
session_file['Cohort'] = session_file['Cohort'].astype(str)
print(session_file['Cohort'].unique())
session_file.head()
```

```python
# Merge with session file
updated_session_file = pd.merge(session_file, all_output_data, on=['Pair', 'Session', 'Sex', 'Cohort'], how='left')
updated_session_file.head(20)
```

```python
# Remove FPS and Dimensions (Miniscope Image)
updated_session_file.drop(columns=['FPS', 'Dimensions'], inplace=True)
updated_session_file.head()
```

```python
# Convert TimeVideoEnd, ReunionDuration, TimeVictimInsertion, TimeOnsetDoorOpening, TimeVictimHeadCrossing, TimeVictimBodyCrossing, Session length to min
updated_session_file['TimeVideoEnd(s)'] = updated_session_file['TimeVideoEnd'].apply(lambda x: int(str(x).split(':')[0])*60 + int(str(x).split(':')[1]) if x is not np.nan else np.nan)
updated_session_file['ReunionDuration(s)'] = updated_session_file['ReunionDuration'].apply(lambda x: int(str(x).split(':')[0])*60 + int(str(x).split(':')[1]) if x is not np.nan else np.nan)
updated_session_file['TimeVictimInsertion(s)'] = updated_session_file['TimeVictimInsertion'].apply(lambda x: int(str(x).split(':')[0])*60 + int(str(x).split(':')[1]) if x is not np.nan else np.nan)
updated_session_file['TimeOnsetDoorOpening(s)'] = updated_session_file['TimeOnsetDoorOpening'].apply(lambda x: int(str(x).split(':')[0])*60 + int(str(x).split(':')[1]) if x is not np.nan else np.nan)
updated_session_file['TimeVictimHeadCrossing(s)'] = updated_session_file['TimeVictimHeadCrossing'].apply(lambda x: int(str(x).split(':')[0])*60 + int(str(x).split(':')[1]) if x is not np.nan else np.nan)
updated_session_file['TimeVictimBodyCrossing(s)'] = updated_session_file['TimeVictimBodyCrossing'].apply(lambda x: int(str(x).split(':')[0])*60 + int(str(x).split(':')[1]) if x is not np.nan else np.nan)
updated_session_file['Session length(s)'] = updated_session_file['Session length'].apply(lambda x: int(str(x).split(':')[0])*60 + int(str(x).split(':')[1]) if x is not np.nan else np.nan)

# Delete old columns
updated_session_file.drop(columns=['TimeVideoEnd', 'ReunionDuration', 'TimeVictimInsertion', 'TimeOnsetDoorOpening', 'TimeVictimHeadCrossing', 'TimeVictimBodyCrossing', 'Session length'], inplace=True)
```

```python
session_file = pd.read_csv('data/SESSION_FILE_insilico_2024-11-25_13-03-00.csv')
```

```python
updated_session_file = session_file
updated_session_file['Normalized components'] = updated_session_file['Num Accepted'] / updated_session_file['Ca Imaging Length(s)']
#print(data.groupby(['Mouse']).head())
# For each unique mouse, sum the normalized components over all days
sum_normalized_components = updated_session_file.groupby(['Mouse'])['Normalized components'].sum()
# Add the sum of normalized components to the data
updated_session_file = updated_session_file.join(sum_normalized_components, on='Mouse', rsuffix='_sum')
# Compute K_id(Num Accepted components for animal i and day d) as k_id (normalized by Session Duration) /sum(k_il over all days) for each animal i and day d
updated_session_file['Normalized components'] = (updated_session_file['Normalized components'] / updated_session_file['Normalized components_sum']) * 100

# Remove the sum of normalized components
updated_session_file = updated_session_file.drop(columns=['Normalized components_sum'])

# Round Ca Imaging Length(s) and Normalized components to 2 decimal places
updated_session_file['Ca Imaging Length(s)'] = updated_session_file['Ca Imaging Length(s)'].round(2)
updated_session_file['Normalized components'] = updated_session_file['Normalized components'].round(2)
updated_session_file.head()
```

```python
# Save the updated session file
session_file_path = 'data/SESSION_FILE_insilico_2024-11-25_13-03-00.csv'
updated_session_file.to_csv(session_file_path.replace('.csv', '_updated.csv'), index=False)
print(f"Saved updated session file to {session_file_path.replace('.csv', '_updated.csv')}")
```
