```python
order = [1, 2, 3, 4, '5+']
sns.barplot(df, y='99th percentile correlation values (all)', x='DayOfHelping', order=order, label='All', fill=False)
sns.barplot(df, y='99th percentile correlation values (lib -> lib)', x='DayOfHelping', order=order, label='Liberation', fill=False)
sns.barplot(df, y='99th percentile correlation values (ran -> ran)', x='DayOfHelping', order=order, label='Random Event', fill=False)

# Load data
df = pd.read_csv('processed-data/FC_DATA_functional_connectivity_2024-10-21_17-31-54.csv')

# Reshape data for plotting with hue
df_melted = df.melt(id_vars=['Cohort', 'Session', 'Mouse', 'DayOfHelping'], 
                    value_vars=['99th percentile correlation values (all)', 
                                '99th percentile correlation values (lib -> lib)', 
                                '99th percentile correlation values (ran -> ran)'], 
                    var_name='Label', value_name='Percentile Values')

# Plot using hue for different labels
sns.barplot(data=df_melted, x='DayOfHelping', y='Percentile Values', hue='Label', order=[1, 2, 3, 4, '5+'])
plt.show()
```
